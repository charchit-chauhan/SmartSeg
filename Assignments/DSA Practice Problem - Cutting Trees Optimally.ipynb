### Problem Statement:
You are given N trees arranged in a straight line. Each tree has a height H[i]. You are allowed to cut
a tree only if its height is strictly greater than both of its immediate neighbors. When you cut a tree: 
You gain wood equal to its height. - The tree is removed from the line. - The neighbors become
adjacent. Your task is to determine the maximum total wood you can collect.

```python
N = int(input())
arr = list(map(int, input().split()))

total_wood = 0

while True:
    cut_made = False
    
    # Check from 1 to len(arr)-2 (because first and last cannot be cut)
    for i in range(1, len(arr) - 1):
        if arr[i] > arr[i - 1] and arr[i] > arr[i + 1]:
            total_wood += arr[i]
            arr.pop(i)   # Remove tree
            cut_made = True
            break        # Restart checking
    
    if not cut_made:
        break

print(total_wood)
```
