### Problem 1: Find All Multiples of 3

You are given an array A of N integers. Return all numbers from the array that are divisible by 3. If
no such number exists, print -1.

```python

N = int(input())
arr = list(map(int, input().split()))


result = []


for num in arr:
    if num % 3 == 0:
        result.append(num)


if len(result) == 0:
    print(-1)
else:
    print(*result)
```

### Problem 2: Longest Consecutive Multiples of 3
You are given an array A of N integers. Find the length of the longest consecutive sequence
(continuous in array) where all elements are divisible by 3.

```python

N = int(input())
arr = list(map(int, input().split()))

current_length = 0
max_length = 0

for num in arr:
    if num % 3 == 0:
        current_length += 1
        max_length = max(max_length, current_length)
    else:
        current_length = 0

print(max_length)
```
