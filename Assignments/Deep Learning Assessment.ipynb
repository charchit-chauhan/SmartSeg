### Question -5

5. What is backpropagation? Explain the role of the chain rule in updating weights. Why is gradient
descent used?

Backpropagation is a method which is  used to reduce the error.

Steps:

1 Calculate output

2 Find error

3 Send error backward

4 Update weights

### Role of Chain Rule

Chain rule helps  to calculate how much each weight is responsible for  the errors.

### Why Gradient Descent?

Gradient Descent helps  to find minimum  the error by updating  their weights step-by-step.

Overfitting means  the model  learns training data too well but it is  performs badly on new data.

Regularization –  It adds penalty to large weights

Early Stopping – It Stop training early

More Data – It increase the training data
