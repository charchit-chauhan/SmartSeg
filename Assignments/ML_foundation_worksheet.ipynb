                                                  Section 1: Markov Decision Processes (MDP)



1. Define a Markov Decision Process (MDP).

   MDP is a decision making model where outcomes depend on actions and chance over time

2. List and explain the five components of an MDP.

   States : A set of all possible situations the agent can be in.

   Actions : All possible moves the agent can make.

   Transitions : The probability of moving to a one state to another given a specific action.

   Rewards: feedback received after an action.

   Discount Factor: A value between 0 and 1 that determines the importance of future rewards.

3. Explain the Markov Property.

   Markov property can  be simply  that the future  depend on present state, not the past.

4. Difference between policy vs value function, and reward vs return.

   Policy :- The  mapping that tells the agent which action to taken in a given state.

   Value Function :- An estimate of how good a state is in terms of expected future rewards.

   Reward: The immediate gain from one action.

   Return: The total accumulated  reward over time.

                                                Section 2: Python Programming Basics

1. Generate numbers from 1 to 20 and extract even numbers using a loop and list comprehension.

```python

#Code 
#1 Loop method
even_no = []
for i in range(1, 21):
    if i % 2 == 0:
        even_no.append(i)

print("Even numbers :", even_no)

#LIST COMPREHENSION METHOD
even_no = [i for i in range(1, 21) if i % 2 == 0]

print("Even numbers :", even_no)
```

2. Write a function to compute the mean of a list.

```python
#CODE

def mean(no):
    if not no:
        return 0  # Avoid division by zero
    return sum(no) / len(no)

# Example usage:
data = [10, 20, 30, 40, 50]
value = mean(data)
print("Mean of the list:", value)
```

3. Create NumPy arrays and compute statistics.

```python
#CODE

import numpy as np


data = np.array([12, 15, 20, 22, 18, 30, 25])


mean = np.mean(data)
median = np.median(data)
std_no = np.std(data)
min_no = np.min(data)
max_no = np.max(data)
sum_no = np.sum(data)


print("Data:", data)
print("Mean:", mean)
print("Median:", median)
print("Standard Deviation:", std_no)
print("Minimum:", min_no)
print("Maximum:", max_no)
print("Sum:", sum_no)
```

                                                         Section 3: Decision Trees

1. What problems do decision trees solve?

   Decision trees are used for both Classification and Regression 

2. Define entropy and information gain.

   Entropy: A measure of disorder or uncertainty in a dataset.

   Information Gain: The reduction in entropy achieved by splitting the data on a specific feature.

3. Why do decision trees overfit?

   This happens when the tree becomes too deep and complex, essentially "memorizing" the training data instead of finding general patterns

4. Given a small dataset, decide the best root node intuitively.
   Python Practice: Train a DecisionTreeClassifier with max_depth=3 and generate predictions

```python
#CODE
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder
import pandas as pd


data = {
    'Weather': ['Sunny', 'Rainy', 'Overcast', 'Sunny'],
    'Temperature': ['Hot', 'Cool', 'Mild', 'Hot'],
    'Play': ['No', 'Yes', 'Yes', 'No']
}


df = pd.DataFrame(data)


weather = LabelEncoder()
temp = LabelEncoder()
play = LabelEncoder()

df['Weather'] = weather.fit_transform(df['Weather'])
df['Temperature'] = temp.fit_transform(df['Temperature'])
df['Play'] = play.fit_transform(df['Play'])

X = df[['Weather', 'Temperature']]
y = df['Play']


model = DecisionTreeClassifier(max_depth=3)
model.fit(X, y)


predictions = model.predict(X)
print("Predictions:", predictions)
```

                                                     Section 4: K-Means Clustering

1. Is K-Means supervised or unsupervised?

K-Means is an Unsupervised learning algorithm

2. Describe the K-Means algorithm steps.

1-Pick k random starting points (centroids).

2-Assign each data point to the nearest centroid.

3-Move the centroids to the center of the assigned points.

4-Repeat until the centroids stop moving.

3. Why is feature scaling important?

It is important because K-Means uses Euclidean distance. If one feature has much larger numbers than another, it will dominate the distance calculation unfairly

4. Explain the Elbow Method and inertia.

A technique to find the best number of clusters ($k$) by plotting Inertia  and looking for the "elbow" where the improvement slows down

Python Practice: 
Fit a KMeans model with k=3 and extract cluster labels.

```python
#CODE

from sklearn.cluster import KMeans
import numpy as np

X = np.array([[1, 2], [1, 4], [1, 0],
              [10, 2], [10, 4], [10, 0]])

kmeans = KMeans(n_clusters=3, random_state=0).fit(X)
labels = kmeans.labels_
print(labels)
```
